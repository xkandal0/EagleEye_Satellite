```python
# COMMS: Payload Data Budget and Downlink Requirements
# EagleEye-1 Mission

import numpy as np

# 1. PAYLOAD REQUIREMENTS [REQ-PAY-020]
gsd_m = 50.0        # Ground Sampling Distance (meters/pixel)
swath_km = 100.0    # Observation swath width on the ground
bit_depth = 12      # 12-bit ADC (Standard for MWIR thermal sensors)

# 2. ORBITAL PARAMETERS (From previous simulations)
sunlight_min = 58.85
pass_duration_min = 7.38

# 3. DATA GENERATION MODEL
# We need overlap between pictures for continuous mapping. 
# At 7.6 km/s, taking a picture every 5 seconds is standard for this swath.
images_per_orbit = (sunlight_min * 60) / 5.0

# Calculate pixels per image (Assuming a square sensor matrix)
pixels_per_line = (swath_km * 1000) / gsd_m
total_pixels = pixels_per_line**2

# Image size in Megabytes (MB)
image_size_bits = total_pixels * bit_depth
image_size_mb = image_size_bits / (8 * 1024 * 1024)

# Total data generated in one orbit
total_data_mb_per_orbit = image_size_mb * images_per_orbit

# 4. DOWNLINK CAPABILITY REQUIREMENT
pass_duration_sec = pass_duration_min * 60.0

# Required Data Rate (Mbps - Megabits per second) to empty the memory in one pass
# Formula: (Total Data in Megabits) / (Pass Time in Seconds)
required_data_rate_mbps = (total_data_mb_per_orbit * 8) / pass_duration_sec

print("=== EAGLEEYE-1 MISSION DATA BUDGET ===")
print(f"Sensor Matrix:       {pixels_per_line:.0f} x {pixels_per_line:.0f} pixels")
print(f"Raw Image Size:      {image_size_mb:.2f} MB")
print(f"Images per Orbit:    {images_per_orbit:.0f}")
print(f"Total Data / Orbit:  {total_data_mb_per_orbit:.2f} MB")
print("--------------------------------------")
print(f"REQUIRED RADIO LINK: {required_data_rate_mbps:.2f} Mbps")
print("======================================")
```
